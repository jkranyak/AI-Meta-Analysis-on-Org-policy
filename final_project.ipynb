```python
def generate_image(prompt):
    output = replicate.run(
        "stability-ai/stable-diffusion-3",
        input={
            "cfg": 3.5,
            "steps": 28,
            "prompt": prompt,
            "aspect_ratio": "1:1",
            "output_format": "png",
            "output_quality": 90,
            "negative_prompt": "low quality, blurry, pixelated",
            "prompt_strength": 0.85
        }
    )
    print(output)
```
